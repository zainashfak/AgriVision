```python
import numpy as np

# Maximum yield loss (%) a disease can cause at 100% severity.
# These are assumed values, not measured from your data.
MAX_LOSS = {
    "Healthy": 0,
    "YellowRust": 50,
    "BrownRust": 35,
    "Septoria": 30,
    "Mildew": 25,
}

def weather_factor(humidity_pct, rain_mm):
    # humid, rainy seasons make disease spread faster
    h = np.clip(humidity_pct / 60, 0.8, 1.3)
    r = np.clip(1 + (rain_mm - 2) * 0.05, 0.9, 1.2)
    return float(h * r)

def estimate_yield_loss(disease, severity_pct, humidity_pct, rain_mm):
    base = MAX_LOSS[disease] * (severity_pct / 100)
    loss = base * weather_factor(humidity_pct, rain_mm)
    return round(min(loss, 100), 1)

# Example: season 2025-26 averages from notebook 02
humidity, rain = 55.0, 2.14

tests = [("Healthy", 1.0), ("YellowRust", 10.2), ("BrownRust", 5.0), ("Septoria", 20.0)]
for disease, sev in tests:
    loss = estimate_yield_loss(disease, sev, humidity, rain)
    print(f"{disease:<11} severity {sev:>5.1f}%  ->  estimated yield loss {loss}%")
```
